```python
import random
from random import sample
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler, LabelEncoder
from sklearn.linear_model import LinearRegression
import kagglehub
import csv
import os
```

# Regresion

https://www.kaggle.com/datasets/yasserh/advertising-sales-dataset

```python
# Download latest version
path = kagglehub.dataset_download("yasserh/advertising-sales-dataset")

ruta_archivo = os.path.join(path, "Advertising Budget and Sales.csv")

datos = pd.read_csv(ruta_archivo)
```

```python
datos.info()
```

```python
datos.describe()
```

```python
sns.pairplot(datos, x_vars=['TV Ad Budget ($)','Radio Ad Budget ($)','Newspaper Ad Budget ($)'], y_vars='Sales ($)', size=7, aspect=0.8,kind = 'reg')
```

```python
# Calcular la matriz de correlación
correlation_matrix = datos.drop(columns=['Unnamed: 0']).corr()

# Visualizar el mapa de correlaciones con seaborn
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Mapa de Correlaciones")
plt.show()
```

```python
X_independientes = datos.iloc[:,1:4]
X_independientes
```

```python
Y_dependiente = datos.iloc[:,4]
Y_dependiente
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X_independientes, Y_dependiente,train_size=.80,  random_state=1279)
```

```python
modelo_rm = LinearRegression() 
modelo_rm.fit(X_train,Y_train)
```

```python
# Coeficientes
print("Coeficientes:", modelo_rm.coef_)

# Intercepto
print("Intercepto:", modelo_rm.intercept_)

# R-cuadrado
r_cuadrado = modelo_rm.score(X_test, Y_test)
print("R-cuadrado:", r_cuadrado)
```

```python
y_pred=modelo_rm.predict(X_test)
```

```python
y_pred
```

```python
import matplotlib.pyplot as plt
plt.scatter(Y_test, y_pred)
plt.xlabel("Valores reales")
plt.ylabel("Predicciones")
plt.title("Valores reales vs. Predicciones")
plt.show()
```

```python
residuos = Y_test - y_pred
plt.scatter(y_pred, residuos)
plt.xlabel("Predicciones")
plt.ylabel("Residuos")
plt.axhline(y=0, color='r', linestyle='--')
plt.title("Gráfico de Residuos")
plt.show()
```

```python
plt.hist(residuos, bins=20)
plt.xlabel("Residuos")
plt.ylabel("Frecuencia")
plt.title("Histograma de Residuos")
plt.show()
```
